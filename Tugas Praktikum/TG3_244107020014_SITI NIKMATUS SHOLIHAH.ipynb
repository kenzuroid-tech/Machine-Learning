```python
import pandas as pd

dpath = '../Data Praktikum/wbc.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
df_simpan = df['diagnosis']

df_drop = df.drop(columns=['id', 'diagnosis', 'Unnamed: 32'])
df_drop.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_simpan_encoded = le.fit_transform(df['diagnosis'])
```

```python
df_simpan_encoded[:5]
```

```python
from sklearn.preprocessing import StandardScaler

std = StandardScaler()
df_drop_std = std.fit_transform(df_drop)
```

```python
df_drop_std[:3, :30]
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd

# Split pertama: 80% train, 20% unseen
df_drop_train, df_drop_unseen, df_simpan_train, df_simpan_unseen = train_test_split(
    df_drop_std,
    df_simpan_encoded,
    test_size=0.2,
    random_state=0,
    stratify=df_simpan_encoded
)

# Split kedua: bagi unseen jadi val dan test
df_drop_val, df_drop_test, df_simpan_val, df_simpan_test = train_test_split(
    df_drop_unseen,
    df_simpan_unseen,
    test_size=0.5,
    random_state=0,
    stratify=df_simpan_unseen
)

# Ubah numpy array jadi pandas Series
df_simpan_encoded_series = pd.Series(df_simpan_encoded, name='diagnosis')
df_simpan_train_series = pd.Series(df_simpan_train, name='diagnosis')
df_simpan_val_series = pd.Series(df_simpan_val, name='diagnosis')
df_simpan_test_series = pd.Series(df_simpan_test, name='diagnosis')

print('UKURAN DATA')
print(f'Jumlah data asli: {len(df_simpan_encoded)}')
print(f'Jumlah data train: {len(df_simpan_train)}')
print(f'Jumlah data val: {len(df_simpan_val)}')
print(f'Jumlah data test: {len(df_simpan_test)}')

print('DISTRIBUSI LABEL')
print(f'\nJumlah label data asli:\n{df_simpan_encoded_series.value_counts()}')
print(f'\nJumlah label data train:\n{df_simpan_train_series.value_counts()}')
print(f'\nJumlah label data val:\n{df_simpan_val_series.value_counts()}')
print(f'\nJumlah label data test:\n{df_simpan_test_series.value_counts()}')
```
