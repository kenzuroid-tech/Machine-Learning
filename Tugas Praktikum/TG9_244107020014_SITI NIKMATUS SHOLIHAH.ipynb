```python
# Load Data
import pandas as pd
data = pd.read_csv('../Data Praktikum/voice.csv')

data.head()
```

```python
data.info()
data.describe()
data['label'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(data, hue='label')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fitur = data.drop(columns='label').columns

fig, axes = plt.subplots(5, 4, figsize=(16, 18))
for ax, col in zip(axes.flat, fitur):
    sns.boxplot(data=data, x='label', y=col, ax=ax)
    ax.set_title(col)
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(5, 4, figsize=(16, 18))
for ax, col in zip(axes.flat, fitur):
    sns.histplot(data=data, x=col, hue='label', kde=True, ax=ax, legend=(col == fitur[0]))
    ax.set_title(col)
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 12))
sns.heatmap(data.drop(columns='label').corr(), annot=True, fmt='.2f',
            annot_kws={'size': 7}, cmap='coolwarm', center=0)
plt.title('Korelasi Antarfitur')
plt.show()
```
